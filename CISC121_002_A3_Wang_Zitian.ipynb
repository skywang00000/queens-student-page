# Problem 2 (Queues) Number of Students Unable to Eat Lunch

### 1. Understanding & Breakdown

This problem requires calculating the number of students who cannot get a meal. First, the student queue is abstracted as a circular queue, with the sandwich stack maintaining its original order. The core algorithm is to traverse the sandwich stack, checking the preference of the student at the front of the queue each time: if a match is found, the student leaves the queue and the next sandwich is processed; otherwise, the student is moved to the back of the queue and continues to wait. The key point is deadlock detection—if no student matches the sandwich at the top of the stack after a complete traversal of the current queue, it means that the remaining students cannot get a meal, and the process is immediately terminated, returning the number of students remaining in the queue.

```python
def queue1(students: list[int], sandwich: list[int]) -> int:
    count = 0

    while students and sandwich:

        if students[0] == sandwich[0]:
          sandwich.pop(0)
          students.pop(0)
        else:
          students.append(students.pop(0))
          count += 1
        if count > len(students):
            break
        print(len(students))

queue1([1,1,1,0], [1,0,0,0,1,1])
```

# Problem 3 (Recursion) Power of Three

------

### 1. Understanding & Breakdown

This problem asks us to determine if a number is a power of 3. First, we need to determine that numbers less than or equal to zero cannot be powers of 3. Therefore, we need to add a condition. We use a while loop to repeatedly divide by 3 until the result equals 1. If it fails, the result should be false.

### 2. Peer Review & Draft Code

I visited someone's solution: https://leetcode.com/problems/power-of-three/solutions/7075222/one-line-solution-by-mikposp-bm56/?envType=problem-list-v2&envId=recursion. His code is so simple, and it is difficult to understand. I added all the conditions so that it would be easy to understand.

```python
def power3(n: int) -> bool:
    if n <= 0:
        return False
    elif n == 1:
        return True
    else:
        while n > 1:
            if n % 3 != 0:
                return False
            else:
                n = n // 3
        return n == 1
n = int(input("Enter a number: "))
print(power3(n))
```

### 3. Final Correct Code

```python
def power3(n: int) -> bool:
    if n <= 0:
        return False
    else:
        while n % 3 == 0:
                n = n // 3
        return n == 1
n = int(input("Enter a number: "))
print(power3(n))
```

# Problem 4 (Search) Find First and Last Position of Element in Sorted Array
--------

### 1. Understanding & Breakdown

This problem asks us to find a target number in a list, with the condition that the solution should be in logarithmic time complexity. This requires using the binary search method. This method is the most convenient and fastest search method. The only drawback is that the list must be sorted. However, this problem has more than one target; we need to find both the initial and last target positions. If a target is not found, we should return the result in the form [-1, -1].

### 2. Peer Review & Draft Code

I review the code from:https://leetcode.com/problems/find-first-and-last-position-of-element-in-sorted-array/solutions/6779462/beats-100-binary-search-easiest-solution-xz84/. This code use the binary search to solve this question.However, there are too many elements in the function. It is too difficult to understand, and the time will be cost a lot.

```python
def binary_search(numbers: list[int], target: int) -> list[int]:
    left = 0
    right = len(numbers) - 1
    while left <= right:
        mid = (left + right) // 2
        if numbers[mid] == target:
            return [mid-1, mid]
        elif numbers[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return [-1, -1]

numbers = [5,7,7,8,8,10]
target = 8
result = binary_search(numbers, target)
print(result)
```

# Problem 5 (Sorting) Sort Colors

----

### 1. Understanding & Breakdown

This problem requires us to replace colors with numbers and arrange the colors in order. The condition is that we cannot use Python's built-in sored method to solve this. This problem can be solved using a loop, comparing each number in the list starting from the zeroth position with its next smallest element. If the first element is smaller than the zeroth, then the first and zeroth elements are swapped. This process is continued until the end.

### 2. Peer Review & Draft Code

I reviewed a code from:https://leetcode.com/problems/sort-colors/solutions/6843641/video-2-solutions-with-frequency-or-3-po-tw2i/. His code used a seperated for loop to figure out this problem. Furthermore, this function uses a dictionary to record how many of each color are in the list before sorting them. This method is very time-consuming.

```python
def color_sort(nums: list[int]) -> list[int]:
    low = 0
    mid = 0
    high = len(nums) - 1
    for _ in range(len(nums)):
        if mid > high:
            break
        if nums[mid] == 0:
            nums[low], nums [mid] = nums[mid], nums[low]
            low += 1
            mid += 1
        elif nums[mid] == 1:
            mid += 1
        else:
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1

nums = [2,0,2,1,1,0,1,2,1,0]
color_sort(nums)
print(nums)
```

### 3. Final Correct Code

```python
def color_sort(nums: list[int]) -> list[int]:
    low = 0
    mid = 0
    high = len(nums) - 1
    while mid <= high:
        if nums[mid] == 0:
            nums[low], nums [mid] = nums[mid], nums[low]
            low += 1
            mid += 1
        elif nums[mid] == 1:
            mid += 1
        else:
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1

nums = [2,0,2,1,1,0,1,2,1,0]
color_sort(nums)
print(nums)
```
